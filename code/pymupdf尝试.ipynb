```python
from collections import Counter

def most_frequent_number(lst):
    # 创建一个Counter对象，统计列表中每个数字出现的次数
    c = Counter(lst)
    # 使用most_common()方法，返回一个列表，其中包含出现次数最多的数字和它的次数
    most_common = c.most_common(1)
    # 返回出现次数最多的数字，它是most_common列表中第一个元素的第一个元素
    return most_common[0][0]

```

```python
from collections import Counter
def judge_main_text(page):
    """
    思路:找文章中单页中出现front size最多的
    参数:page是单个fitz类对象，且为整篇文章
    返回值:返回出现最多的front size,以及当前page的span大小list
    """
    font_size =[]
    text_dict = page.get_text("dict")  # 获取文本信息字典
    for block in text_dict["blocks"]:  # 遍历每个文本块
        if block["type"] == 0:  # 如果是文字类型
            for line in block["lines"]:  # 遍历每一行
                for span in line["spans"]:  # 遍历每个span
#                     print(span["size"])
                    font_size.append(span["size"])
    return most_frequent_number(font_size),font_size

import fitz
path = r"./media/Financial Sector Taxation.pdf"
doc = fitz.open(path)

record_most_account = []
for i in range(doc.page_count):
    MostFrequentNumber,_ = judge_main_text(doc[i])
    record_most_account.append(MostFrequentNumber)
#正文文本选择要设定一个阈值（经验判断可为-+2）
print(record_most_account)
```
